## Задача 1 (2 б.): Свертка

Докажите, что
$$
\int\Phi(\lambda a) N(a|\mu, \Sigma) = \Phi(\frac{\mu}{\sqrt{\lambda^{-2} + \sigma^2}})
$$

У интеграла в левой части не указана интегрируемая переменная и пределы интегрирования. Предположим, что имелось в виду следующее выражение:
$$ \int_{-\infty}^{+\infty} \Phi(\lambda a) N(a|\mu, \Sigma) da $$

Также предположим, что $\Sigma = \sigma^2$ и $\mu, \lambda, \sigma \gt 0$, так как в задании "не обработаны" случаи деления на ноль, извлечения корня из отрицательных чисел.

По определению,
$$ \Phi(\lambda a) = \int_{-\infty}^{\lambda a} N(t|\mu = 0, \Sigma = 1) dt $$

Введем индикаторную функцию и заменим пределы интеграла:
$$ \int_{-\infty}^{\lambda a} N(t|\mu = 0, \Sigma = 1) dt = \int_{-\infty}^{+\infty} \mathbf 1\{t \le \lambda a\} N(t|\mu = 0, \Sigma = 1) dt $$

Тогда, исходное выражение:
$$ \int_{-\infty}^{+\infty} \Phi(\lambda a) N(a|\mu, \Sigma) da = \int_{-\infty}^{+\infty} \left[ \int_{-\infty}^{+\infty} \mathbf 1\{t \le \lambda a\} N(t|\mu = 0, \Sigma = 1) dt \right] N(a|\mu, \Sigma) da $$

Приводим к двойному интегралу:
$$ \int_{-\infty}^{+\infty} \int_{-\infty}^{+\infty} \mathbf 1\{t \le \lambda a\} N(t|\mu = 0, \Sigma = 1) N(a|\mu, \Sigma) dt da $$

Пусть $X \sim N(0, 1)$, $Y \sim N(\mu, \Sigma)$ - независимые случайные величины. Пусть $\phi_X(x)$ и $\phi_Y(x)$ - их плотности распределения. 

Тогда двойной интеграл экивалентен
$$ \int_{-\infty}^{+\infty} \int_{-\infty}^{+\infty} \mathbf 1\{t \le \lambda a\} \phi_X(t) \phi_Y(a) dt da $$

$\phi_X(t) \phi_Y(a)$ - совместная плотность независимых $\left(X, Y\right)$. По определению непрерывной вероятности,
$$ \int_{-\infty}^{+\infty} \int_{-\infty}^{+\infty} \mathbf 1\{t \le \lambda a\} \phi_X(t) \phi_Y(a) dt da = \mathbb P(X\le \lambda Y) $$

Преобразуем вероятность:
$$\mathbb P(X\le \lambda Y) = \mathbb P(\lambda^{-1} X \le Y) = \mathbb P(0 \le Y - \lambda^{-1} X) =   \mathbb P(Y - \lambda^{-1} X \gt 0)$$

Пусть $B = Y - \lambda^{-1} X$. По свойствам нормального распределения:
$$X \sim N(0, 1) \Rightarrow \lambda^{-1} X \sim N(0, \lambda^{-2})$$
$$Y \sim N(\mu, \Sigma) \Rightarrow B \sim N(\mu + 0, \Sigma + \lambda^{-2})$$

Таким образом, левая часть исходного уравнения равна $\mathbb P(B \gt 0)$ для $B \sim N(\mu, \Sigma + \lambda^{-2})$

Найдем $\mathbb P(B \gt 0)$ стандартным способом. Пусть $C = \dfrac{B - \mu}{\sqrt{\Sigma + \lambda^{-2}}} \sim N(0, 1)$. Тогда:
$$B = \sqrt{\Sigma + \lambda^{-2}} C + \mu$$
$$\mathbb P(B \gt 0) = \mathbb P(\sqrt{\Sigma + \lambda^{-2}} C + \mu \gt 0) = \mathbb P(C \gt \dfrac{-\mu}{\sqrt{\Sigma + \lambda^{-2}}})$$

Так как $C \sim N(0, 1)$,
$$\mathbb P(C \gt \dfrac{-\mu}{\sqrt{\Sigma + \lambda^{-2}}}) = \mathbb P(C \lt \dfrac{\mu}{\sqrt{\Sigma + \lambda^{-2}}}) = \Phi(\frac{\mu}{\sqrt{\lambda^{-2} + \sigma^2}})$$

Таким образом,
$$ \int_{-\infty}^{+\infty} \Phi(\lambda a) N(a|\mu, \Sigma) da = \mathbb P(B \gt 0) = \Phi(\frac{\mu}{\sqrt{\lambda^{-2} + \sigma^2}})$$

## Задача 2 (3 б.): Метод Ньютона

Метод Ньютона - итеративный алгоритм для поиска корней функции $f(x)$:

$x_{new} = x_{old} - J_f(x_{old})^{-1} f(x_{old})$

где $J$ - матрица Якоби. Выведите из этого определения формулу, которую мы использовали для поиска моды $\mathbb{P}(f|y)$:

$$
f^{new} = K(I+WK)^{-1} \cdot \Big[y-\sigma(f)+Wf\Big]
$$

Для поиска $f$ необходимо максимизировать апостериорную плотность $L(f | y)$.

По Байесу, $L(f | y) = L(y | f) L(f)$. Для удобства, будем рассматривать $\log L(f | y) = \log L(y | f) + \log L(f)$

(1) Функция правдоподобия:
$$L(y | f) = \prod_{i = 1}^{n} p(y_i | f_i) = \prod_{i = 1}^{n} \sigma(f_i)^{y_i} (1 - \sigma(f_i)) ^ {1 - y_i}$$
$$\log L(y | f) = \sum_{i = 1}^{n} y_i \log \sigma(f_i) + (1 - y_i) \log (1 - \sigma(f_i))$$
$$\log L(y | f) = \sum_{i = 1}^{n} y_i \left(f_i - \log (1 + e^{f_i})\right) + (1 - y_i) (-\log (1 + e^{f_i}))$$
$$\log L(y | f) = \sum_{i = 1}^{n} y_i f_i - y_i \log (1 + e^{f_i}) -\log (1 + e^{f_i}) + y_i \log (1 + e^{f_i})$$
$$\log L(y | f) = \sum_{i = 1}^{n} y_i f_i -\log (1 + e^{f_i})$$

(2) Априорная плотность:
$$ L(f) = N(f | 0, K) = \dfrac{1}{(2 \pi)^{\frac{n}{2}} \sqrt{|K|}} e^{-\frac{1}{2} f^T K^{-1} f} $$
$$ \log L(f) = - \log (2 \pi)^{\frac{n}{2}} - \log \sqrt{|K|} - \frac{1}{2} f^T K^{-1} f $$
$$ \log L(f) = - \dfrac{n}{2} \log (2 \pi) - \dfrac{1}{2} \log |K| - \frac{1}{2} f^T K^{-1} f $$

Соответственно,
$$\log L(f | y) = \sum_{i = 1}^{n} \left( y_i f_i -\log (1 + e^{f_i}) \right) - \dfrac{n}{2} \log (2 \pi) - \dfrac{1}{2} \log |K| - \frac{1}{2} f^T K^{-1} f $$

Производная:
$$ \log L(f | y)_{f_i}^{'} = y_i - \sigma(f_i) - (K^{-1} f)_{i} $$

Градиент:
$$ \nabla L(f | y) (f) = y - \sigma(f) - K^{-1} f $$

Условие максимума $L(f | y)$: $\nabla L(f | y) (f) = 0$. Таким образом, задача сводится к поиску корней функции $\nabla L(f | y) (f)$

В терминах метода Ньютона:
$$ x := f \Rightarrow x_{new} := f_{new}, x_{old} := f_{old}$$
$$ f(x_{old}) := \nabla L(f | y) (f) $$
$$ J_f(x_{old}) := \nabla^2 L(f | y) (f) $$

Матрица Якоби:
$$ J_f(x_{old}) = \nabla^2 L(f | y) (f) = \nabla \left[ \nabla L(f | y) (f) \right] (f) = \nabla \left[ y - \sigma(f) - K^{-1} f \right] (f) = -W - K^{-1}$$

Обратная к матрцие Якоби:
$$(W + K^{-1})^{-1} = ((WK + I) K^{-1})^{-1} = K (WK + I)^{-1} = K (I + WK)^{-1}$$

Шаг метода Ньютона:
$$f_{new} = f - \left( -W - K^{-1} \right)^{-1} \left(y - \sigma(f) - K^{-1} f\right)$$
$$f_{new} = f + K (I + WK)^{-1} \left(y - \sigma(f) - K^{-1} f\right)$$
$$f_{new} = f + K (I + WK)^{-1} \left(y - \sigma(f) \right) - K (I + WK)^{-1} K^{-1} f$$
$$f_{new} = K (I + WK)^{-1} \left(y - \sigma(f) \right) + \left[ \underbrace{I} - K (I + WK)^{-1} K^{-1} \right] f$$
$$f_{new} = K (I + WK)^{-1} \left(y - \sigma(f) \right) + \left[ K \underbrace{I} K^{-1} - K (I + WK)^{-1} K^{-1} \right] f$$
$$f_{new} = K (I + WK)^{-1} \left(y - \sigma(f) \right) + \left[ K (I + WK)^{-1} (I + WK) K^{-1} - K (I + WK)^{-1} K^{-1} \right] f$$
$$f_{new} = K (I + WK)^{-1} \left(y - \sigma(f) \right) + \left[ \left(K (I + WK)^{-1}\right) \left((I + WK) K^{-1} - K^{-1}\right) \right] f$$
$$f_{new} = K (I + WK)^{-1} \left(y - \sigma(f) \right) + \left[ \left(K (I + WK)^{-1}\right) (\underbrace{I} + WK \underbrace{- I}) K^{-1} \right] f$$
$$f_{new} = K (I + WK)^{-1} \left(y - \sigma(f) \right) + \left[ K (I + WK)^{-1} W \underbrace{K K^{-1}}_{I} \right] f$$
$$f_{new} = K (I + WK)^{-1} \left(y - \sigma(f) \right) + K (I + WK)^{-1} W f$$
$$f_{new} = \overbrace{K (I + WK)^{-1}} \left(y - \sigma(f) + Wf\right)$$
